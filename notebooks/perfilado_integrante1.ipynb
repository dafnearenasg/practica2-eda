```python
import pandas as pd
import numpy as np
```

```python
# Nombres oficiales de las columnas para Adult Income
nombres_adult = ['age', 'workclass', 'fnlwgt', 'education', 'education-num', 
                 'marital-status', 'occupation', 'relationship', 'race', 'sex', 
                 'capital-gain', 'capital-loss', 'hours-per-week', 'native-country', 'income']

# Cargar Adult indicando que NO tiene encabezado y pasándole la lista de nombres
adult = pd.read_csv('../data/adult.csv', skipinitialspace=True, na_values='?', 
                    header=None, names=nombres_adult)

telco = pd.read_csv('../data/telco_churn.csv')

# Verificar vacíos en Telco antes de convertir
print("Valores vacíos ocultos en TotalCharges:", (telco['TotalCharges'].astype(str).str.strip() == '').sum())
telco['TotalCharges'] = pd.to_numeric(telco['TotalCharges'], errors='coerce')

# Imprimir dimensiones y tipos de datos
for nombre, df in [("adult", adult), ("telco", telco)]:
    print(f"--- {nombre.upper()} ---")
    print(f"Dimensiones (filas, columnas): {df.shape}")
    df.info()
    print("\n")
```

```python
def perfil(df):
    return pd.DataFrame({
        'tipo': df.dtypes.astype(str),
        'ausentes': df.isna().sum(),
        'pct_ausentes': (df.isna().mean() * 100).round(2),
        'unicos': df.nunique(dropna=True),
    })

print("PERFIL - ADULT INCOME")
display(perfil(adult))

print("\nPERFIL - TELCO CHURN")
display(perfil(telco))
```

```python
def estadisticos(df):
    num = df.select_dtypes('number')
    d = num.describe(percentiles=[.01, .05, .25, .5, .75, .95, .99]).T
    d['mediana'] = num.median()
    d['asimetria'] = num.skew()
    return d.round(2)

print("--- ESTADÍSTICOS ADULT INCOME ---")
display(estadisticos(adult))

print("\n--- ESTADÍSTICOS TELCO CHURN ---")
display(estadisticos(telco))
```

```python
def categoricas(df):
    cat = df.select_dtypes(exclude='number')
    for c in cat.columns:
        print(c, "| moda:", cat[c].mode()[0])
        print(cat[c].value_counts(dropna=False).to_string(), "\n")

categoricas(adult)
```

```python
def inconsistencias(df):
    print('Duplicados:', df.duplicated().sum())
    print("Constantes:", [c for c in df if df[c].nunique(dropna=False) <= 1])
    print("Casi constantes (>99%):", [c for c in df
          if df[c].value_counts(normalize=True, dropna=False).iloc[0] > 0.99])
    print("Identificadores únicos:", [c for c in df if df[c].nunique() == len(df)])
    cat = df.select_dtypes(exclude="number")
    for c in cat:
        norm = cat[c].dropna().astype(str).str.strip().str.lower()
        if norm.nunique() != cat[c].nunique():
            print("Categorías inconsistentes en:", c)
    corr = df.select_dtypes("number").corr().abs()
    par = corr.where(np.triu(np.ones(corr.shape, bool), 1)).stack()
    print("Correlaciones > 0.8:")
    print(par[par > 0.8])

inconsistencias(adult)
inconsistencias(telco)
```

```python
# Rangos válidos según el significado de cada variable
rangos_adult = {'age': (17, 90), 'hours-per-week': (1, 99)}
rangos_telco = {'tenure': (0, 72), 'MonthlyCharges': (0, 200)}

def fuera_de_rango(df, rangos):
    for col, (lo, hi) in rangos.items():
        # ~ significa "lo contrario a", es decir, los que NO están entre 'lo' y 'hi'
        fuera = (~df[col].between(lo, hi)).sum()
        print(f"Variable '{col}' fuera del rango lógico ({lo} a {hi}): {fuera} registros")

print("--- VALORES FUERA DE RANGO: ADULT INCOME ---")
fuera_de_rango(adult, rangos_adult)

print("\n--- VALORES FUERA DE RANGO: TELCO CHURN ---")
fuera_de_rango(telco, rangos_telco)
```
